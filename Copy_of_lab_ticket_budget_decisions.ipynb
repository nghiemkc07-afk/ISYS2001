c## Your final prompt

> Paste your complete, finished prompt below (double-click to edit).

```
Write a complete Python program that creates a 'Budget Checker' tool. The program should guide the user through inputting necessary values and then provide a clear decision based on those inputs.

Here are the step-by-step requirements:

1.  **Input Collection**:
    *   Prompt the user to enter their 'name'. Use `input()` with the message: 'Please enter your name: '.
    *   Prompt the user to enter their `total monthly budget`. Use `input()` with the message: 'Please enter your total monthly budget: '. Convert this input to a floating-point number.
    *   Prompt the user to enter the `current expense amount`. Use `input()` with the message: 'Please enter the current expense amount: '. Convert this input to a floating-point number.
    *   Prompt the user to enter a `percentage threshold` for a single expense. This threshold defines what percentage of the total budget a single expense should not exceed. Use `input()` with the message: 'Please enter the maximum percentage a single expense should be (e.g., 10 for 10%): '. Convert this input to a floating-point number.

2.  **Calculation**:
    *   Calculate the percentage the `current expense amount` represents of the `total monthly budget`. The formula is: `(current_expense_amount / total_monthly_budget) * 100`.

3.  **Decision Logic and Output**:
    *   Compare the calculated expense percentage with the `percentage threshold`.
    *   **If the calculated expense percentage is greater than the `percentage threshold`**:
        *   Print a warning message. The message should clearly state the current expense amount, the total budget, the calculated percentage, and the threshold that was exceeded. Format the output to two decimal places for monetary values and percentages. Example format: "Warning: This expense ($XXX.XX) is YYY.YY% of your budget ($ZZZ.ZZ), which exceeds your T.TT% limit. Consider if this expense is necessary!"
    *   **If the calculated expense percentage is less than or equal to the `percentage threshold`**:
        *   Print a positive message. The message should clearly state the current expense amount, the total budget, the calculated percentage, and confirm it's within the threshold. Format the output to two decimal places for monetary values and percentages. Example format: "Good news! This expense ($XXX.XX) is YYY.YY% of your budget ($ZZZ.ZZ), which is within your T.TT% limit. Keep up the good work!"

4.  **Error Handling**: Implement basic error handling for non-numeric inputs. If any of the inputs cannot be converted to a float, print an error message: "Invalid input. Please enter a numeric value." and then exit the program gracefully.

5.  **Output Clarity**: Ensure all output messages are user-friendly and easy to understand.

```

````python
try:
    # 1. Input Collection
    user_name = input('Please enter your name: ')
    total_monthly_budget_str = input('Please enter your total monthly budget: ')
    total_monthly_budget = float(total_monthly_budget_str)

    current_expense_amount_str = input('Please enter the current expense amount: ')
    current_expense_amount = float(current_expense_amount_str)

    percentage_threshold_str = input('Please enter the maximum percentage a single expense should be (e.g., 10 for 10%): ')
    percentage_threshold = float(percentage_threshold_str)

    # Basic validation for positive numbers, as budget/expense/threshold should be positive
    if total_monthly_budget <= 0 or percentage_threshold <= 0:
        print("Invalid input. Total monthly budget and percentage threshold must be positive.")
    elif current_expense_amount < 0:
        print("Invalid input. Current expense amount cannot be negative.")
    else:
        # 2. Calculation
        if total_monthly_budget == 0:
            # Handle division by zero if total_monthly_budget is 0 after validation
            print("Error: Total monthly budget cannot be zero for percentage calculation.")
        else:
            calculated_percentage = (current_expense_amount / total_monthly_budget) * 100

            # 3. Decision Logic and Output
            if calculated_percentage > percentage_threshold:
                print(f"Hi {user_name}! Warning: This expense (${current_expense_amount:.2f}) is {calculated_percentage:.2f}% of your budget (${total_monthly_budget:.2f}), which exceeds your {percentage_threshold:.2f}% limit. Consider if this expense is necessary!")
            else:
                print(f"Hi {user_name}! Good news! This expense (${current_expense_amount:.2f}) is {calculated_percentage:.2f}% of your budget (${total_monthly_budget:.2f}), which is within your {percentage_threshold:.2f}% limit. Keep up the good work!")

except ValueError:
    # 4. Error Handling for non-numeric inputs
    print("Invalid input. Please enter a numeric value.")
````
