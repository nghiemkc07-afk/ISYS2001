## Task 1: total sales by region, highest first

Load the file into a DataFrame, clean whatever needs cleaning so the sale values are
real numbers, and produce the total `Total_Sale` for each `Region`, ordered from
highest to lowest.

My plan, in plain words:

1.  **Load Data**: Read the `sales_data.csv` file into a pandas DataFrame.
2.  **Clean 'Quantity' column**: Remove the '$' symbol and convert this column to an integer type.
3.  **Clean 'Unit_Price' and 'Total_Sale' columns**: Remove the '$' symbol and convert these columns to a float type.
4.  **Calculate total sales by region**: Group the DataFrame by the `Region` column and sum the `Total_Sale` for each region.
5.  **Sort Results**: Sort the calculated total sales in descending order (from highest to lowest).
6.  **Display Results**.

```python
# Task 1: total sales by region, highest first.
```

```python
import pandas as pd

# 1. Load the data into a DataFrame
df = pd.read_csv('sales_data.csv')

# 2. Clean 'Quantity' column
df['Quantity'] = df['Quantity'].astype(str).str.replace('$', '', regex=False).astype(int)

# 3. Clean 'Unit_Price' and 'Total_Sale' columns
df['Unit_Price'] = df['Unit_Price'].astype(str).str.replace('$', '', regex=False).astype(float)
df['Total_Sale'] = df['Total_Sale'].astype(str).str.replace('$', '', regex=False).astype(float)

# 4. Calculate total sales by region
total_sales_by_region = df.groupby('Region')['Total_Sale'].sum()

# 5. Sort the results from highest to lowest
total_sales_by_region_sorted = total_sales_by_region.sort_values(ascending=False)

# 6. Display the results
display(total_sales_by_region_sorted)
```

## Task 2: units sold per product

Now find how many units of each product were sold in total, using the `Quantity`

My plan, in plain words:

1.  **Group Data**: Group the existing DataFrame `df` by the `Product` column.
2.  **Sum Quantities**: Calculate the sum of the `Quantity` for each product group.
3.  **Sort Results**: Sort the products by their total quantity sold in descending order.

```python
# Task 2: total units sold per product.
```

```python
# Calculate total units sold per product
total_units_sold_per_product = df.groupby('Product')['Quantity'].sum()

# Sort the results from highest to lowest
total_units_sold_per_product_sorted = total_units_sold_per_product.sort_values(ascending=False)

# Display the results
display(total_units_sold_per_product_sorted)
```
